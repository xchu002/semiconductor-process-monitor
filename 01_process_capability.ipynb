```python
import pandas as pd

df = pd.DataFrame({
    "wafer_id": ["W001", "W002", "W003", "W004", "W005", "W006"],
    "tool_id": ["A", "A", "A", "B", "B", "B"],
    "film_thickness_nm": [99.8, 100.4, 100.1, 101.9, 102.4, 102.1],
    "yield": [0.95, 0.96, 0.94, 0.89, 0.87, 0.88]
})

print(df)
```

```python
tool_stats = df.groupby("tool_id")[["film_thickness_nm", "yield"]].mean()

print(tool_stats)
```

```python
correlation = df["film_thickness_nm"].corr(df["yield"])
print(f"Correlation between film thickness and yield: {correlation:.2f}")
```

```python
USL = 105
LSL = 95

process_std = df["film_thickness_nm"].std()

#potential capability index (Cp) calculation
Cp = (USL - LSL) / (6 * process_std)

print("Cp:", Cp)
```

```python
mean = df["film_thickness_nm"].mean()

#process capability index (Cpk) calculation
Cpu = (USL - mean) / (3 * process_std)
Cpl = (mean - LSL) / (3 * process_std)

Cpk = min(Cpu, Cpl)

print("mean:", mean)
print("Cpu:", Cpu)
print("Cpl:", Cpl)
print("Cpk:", Cpk)

#results show that cpu is lower, and thus the process is more likely to produce products that exceed the upper specification limit (USL).
```
