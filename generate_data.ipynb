```python
import numpy as np
import pandas as pd

np.random.seed(42)

n_wafers = 500
```

```python
wafer_ids = [
    f"W{i:04d}"
    for i in range(1, n_wafers + 1)
]
```

```python
lot_ids = [
    f"L{((i - 1) // 25) + 1:03d}"
    for i in range(1, n_wafers + 1)
]
```

```python
tool_ids = np.random.choice(
    ["A", "B", "C"],
    size=n_wafers,
    p=[0.4, 0.35, 0.25]
)
```

```python
temperature = np.random.normal(
    loc=400,
    scale=3,
    size=n_wafers
)

pressure = np.random.normal(
    loc=50,
    scale=2,
    size=n_wafers
)

deposition_time = np.random.normal(
    loc=60,
    scale=1.5,
    size=n_wafers
)
```

```python
df = pd.DataFrame({
    "wafer_id": wafer_ids,
    "lot_id": lot_ids,
    "tool_id": tool_ids,
    "temperature_C": temperature,
    "pressure_mTorr": pressure,
    "deposition_time_s": deposition_time
})

print(df.head())
```

```python
thickness = 100.0

thickness = (
    thickness
    + 0.08 * (df["temperature_C"] - 400)
)
```

```python
thickness = (
    thickness
    + 0.12 * (df["deposition_time_s"] - 60)
)

thickness = (
    thickness
    + 0.03 * (df["pressure_mTorr"] - 50)
)
```
